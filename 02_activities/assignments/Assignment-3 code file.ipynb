```python

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Load the CSV file
df = pd.read_csv("/Users/ahilk/visualization/02_activities/assignments/Toronto Island Ferry Ticket Counts.csv", low_memory=False)

# Group data by Year and sum counts
group_dataset = df.groupby("Year")[["Redemption Count", "Sales Count"]].sum()

# Create positions for grouped bars
years = group_dataset.index
x = np.arange(len(years))
width = 0.35  # bar width

plt.figure(figsize=(10, 7))

plt.bar(x - width/2, group_dataset["Redemption Count"], width, label="Redemption Count")
plt.bar(x + width/2, group_dataset["Sales Count"], width, label="Sales Count")

plt.legend(loc='center left', bbox_to_anchor=(1, 0.5)) #Move legend outside the graph

#Add labels and title
plt.xlabel("Year")
plt.ylabel("Count (in millions)")
plt.title("Redemption vs Sales by Year")
plt.xticks(x, years)
plt.tight_layout()

plt.show()
```

**Alt text:** Bar chart showing Redemption Count vs Sales Count by Year. Redemption Count is lower than Sales Count each year.
